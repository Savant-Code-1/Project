```python
# KEY PROJECT INSIGHTS

print("=" * 50)
print("GYM MEMBER FITNESS SCORE ANALYSIS")
print("=" * 50)

# Overall fitness score

print("\n1. OVERALL FITNESS PERFORMANCE")

print(
    "Average Fitness Score:",
    round(df["Fitness_Score"].mean(), 2)
)

print(
    "Median Fitness Score:",
    round(df["Fitness_Score"].median(), 2)
)


# Correlation

print("\n2. WORKOUT HABITS")

print(
    "Visits Correlation:",
    round(
        df["Visits_Per_Week"].corr(
            df["Fitness_Score"]
        ), 2
    )
)

print(
    "Session Duration Correlation:",
    round(
        df["Avg_Session_Minutes"].corr(
            df["Fitness_Score"]
        ), 2
    )
)


# Best performing categories

print("\n3. CATEGORY COMPARISON")

for column in [
    "Membership_Type",
    "City",
    "Gender",
    "Workout_Preference"
]:

    group_average = df.groupby(
        column
    )["Fitness_Score"].mean()

    print(
        f"Highest average for {column}:",
        group_average.idxmax(),
        "-",
        round(group_average.max(), 2)
    )





print("=" * 50)
```
